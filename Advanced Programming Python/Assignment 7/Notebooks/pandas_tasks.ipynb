```python
import pandas as pd

# --- Task 3: Create a DataFrame ---
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [20, 22, 19, 23, 21],
    'Score': [85, 78, 92, 60, 88]
}

df = pd.DataFrame(data)
print("DataFrame Created:")
print(df)

# --- Task 4: Data Exploration ---
print("\n--- First 3 rows ---")
print(df.head(3))

print("\n--- Summary Statistics ---")
print(df.describe())

print(f"\nAverage Score: {df['Score'].mean()}")

# --- Task 5: Data Filtering ---
print("\n--- Students older than 20 ---")
print(df[df['Age'] > 20])

print("\n--- Students with score above 80 ---")
print(df[df['Score'] > 80])
```
